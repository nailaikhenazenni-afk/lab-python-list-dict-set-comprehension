````python
products = ('t-shirt', 'mug', 'hat', 'book', 'keychain')
inventory = { }
def initialize_inventory(products):
            inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
            return inventory

````

````python



def update_inventory(customer_orders, inventory):
    inventory = {
        product: count 
        for product, count in inventory.items() 
        if count > 0
    }
    inventory = {
        product: (count - 1 if product in customer_orders else count)
        for product, count in inventory.items()
    }

   

    return update_inventory
````

````python
def print_updated_inventory(inventory):
    print("Updated Inventory:")
    print("\n".join([f"{product}: {quantity}" for product, quantity in inventory.items()]))
````

````python
def get_customer_orders(products):
   
    try:
        orders = int(input("Enter the number of customer orders: "))
    except ValueError:
        print("Error: Please enter a valid integer for the number of orders.")
        return set()

    customer_orders = {
        product for _ in range(orders)
        if (product := input("Enter the product names: ")) in products
    }
    return customer_orders
  
````

````python
def calculate_total_price(customer_orders):
  
    total_price= sum(
        float(input(f"Enter the price of {product}: ")) 
        for product in customer_orders
    )


    return total_price
````

````python
def print_total_price_of_customer_order(total_price):
    print(f"Total price of customer order: {total_price[1]:.1f}")
````

````python
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    total_available_products = len(products)

    if total_products_ordered == 0:
        return 0, 0.0

    percentage_unique_products_ordered = (total_products_ordered / total_available_products) * 100

    order_statistics = (total_products_ordered, percentage_unique_products_ordered)

    return order_statistics
````

````python
def print_order_statistics(order_statistics):
    print("\norder statistics:")
    print(f"total products ordered: {order_statistics[0]}")
    print(f"percentage of unique products ordered: {order_statistics[1]:.2f}%")
````

````python
def manage_customer_orders():

    products_list = ['t-shirt', 'mug', 'hat', 'book', 'keychain']
    inventory = initialize_inventory(products_list)
    customer_orders = get_customer_orders()
    updated_inventory = update_inventory(customer_orders, inventory)
    order_statistics= calculate_order_statistics(customer_orders, products_list)
    print_order_statistics(order_statistics)
    print_updated_inventory(updated_inventory)
    total_price = calculate_total_price(customer_orders)
    print_total_price_of_customer_order(total_price)

    return

````
